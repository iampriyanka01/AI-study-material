```python
## Conversation Example

from openai import OpenAI

# Instantiate an OpenAI Client
client = OpenAI(
    api_key="GEMINI_API_KEY",
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

# 2. Initialize conversation history
messages = [
    {"role": "system", "content": "You are a movie buff, you only reply to what is asked for and in very concise manner with comma separated lists wherever applicable"},
    {"role": "user", "content": "Who directed the movie Inception?"},
    {"role": "assistant", "content": "**Christopher Nolan** directed the movie *Inception*."},
    {"role": "user", "content": "What other movies has he directed?"}
    ]

# 3. Send first request
response = client.chat.completions.create(
    model="gemini-2.5-flash",  # Use Gemini model name
    messages=messages
)
assistant_reply = response.choices[0].message.content

print("Assistant:", assistant_reply)
```

```python
from openai import OpenAI

# Initialize conversation history
messages = [
    {"role": "system",
     "content": "You are a movie buff, you only reply to what is asked for and in very concise manner with comma separated lists wherever applicable"},
]

def ask_question(question, messages):

    # Instantiate an OpenAI Client
    client = OpenAI(
        api_key="GEMINI_API_KEY",
        base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
    )

    messages.append({"role": "user", "content": question})

    # 3. Send first request
    response = client.chat.completions.create(
        model="gemini-2.5-flash",  # Use Gemini model name
        messages=messages
    )
    assistant_reply = response.choices[0].message.content
    messages.append({"role": "assistant", "content": assistant_reply})

    return assistant_reply
```

```python
# Step 1: Import the Gemini SDK
from google import genai


# Step 2: Set up the Gemini API key
client = genai.Client(api_key="GEMINI_API_KEY")


exit_words = {"bye", "exit", "quit", "goodbye", "see you"}


while True:
    user_input = input("Ask Gemini (type 'bye' or 'exit' to quit): ")
    if any(word in user_input.lower() for word in exit_words):
        print("Exiting chat. Goodbye!")
        break
    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=user_input
    )
    print(response.text)
```

```python
# Step 1: Import the Gemini SDK
from google import genai

# Step 2: Set up the Gemini API key
client = genai.Client(api_key="GEMINI_API_KEY")

# Create a new chat session using Gemini's chat API
chat = client.chats.create(model="gemini-2.5-flash")
```

```python
response = chat.send_message("Who directed the movie Inception?")
print(response.text)
```

```python
response = chat.send_message("What other movies has he directed?")
print(response.text)
```

```python
print("\nConversation History:")
for message in chat.get_history():
    print(f'{message.role.capitalize()}: {message.parts[0].text}')
```
